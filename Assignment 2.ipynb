```python
'''Write a python program to find the largest of three numbers.'''
numbers = []
a = int(input("enter number 1:"))
b = int(input("enter number 2:"))
c = int(input("enter number 3:"))
numbers.append(a)
numbers.append(b)
numbers.append(c)
print(numbers)
```

```python
print("largest number =",max(numbers))
```
