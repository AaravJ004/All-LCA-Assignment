```python
s1 = (184,"Aarav","CSE",89)
s2 =(195,"Madhav","ECE",89)
s3 = (175,"Sanjit","ECE",90)
students = {184: s1,195: s2,175: s3}
rollno = [184,195,175]
```

```python
'''1.Add a new student record'''
s4 = (165,"Archit","CSE",90)
students[165] = s4
rollno.append(165)
print(students)
```

```python
'''2.Delete an existing student record'''
del students[175]
rollno.remove(175)
print(students)
```

```python
'''3.Update the final details of a student'''
students[195] = (195,"Madhav","ECE",95)
```

```python
'''4.Print final student record'''
print("final students record:",students)
```
