```python
import pandas as pd
import numpy as np
#Read the CSV or Excel file into a Pandas DataFrame.
df=pd.read_excel(r"C:\Users\hp_Pc\Downloads\array.xlsx")
print(df)
```

```python
#Display the first few records
print("First five records only:")
print(df.head())
```

```python
#Display information about the dataset.
print("dataset information:")
print(df.info())
```

```python
#Check for missing values.
print("Total missing values:", df.isnull().sum().sum())
```

```python
#Handle missing values using an appropriate method.
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
#Identify and remove duplicate records.
print("Number of duplicate records:", df.duplicated().sum())
```

```python
#Convert data types wherever required.
df['Name'] = df['Name'].astype('string')
```

```python
#Perform basic data cleaning.
df.columns = df.columns.str.strip()
```

```python
#Display the preprocessed dataset.
display(df)
```

```python
#Display the final dataset information.
print("Final Dataset Information")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```
