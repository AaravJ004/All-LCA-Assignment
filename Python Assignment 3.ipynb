```python
'''Write a python program that accepts the length of three sides of a triangle as inputs.The program should indicate whether or not the triangle is a 
right-angled triangle using function'''
H = int(input("enter value of hypotenuse ="))
P = int(input("enter value of perpendicular ="))
B = int(input("enter value of base ="))
H2 = (H*H)
P2 = (P*P)
B2 = (B*B)
SP2B2 = (P2+B2)
if H2 == SP2B2:
    print("Triangle is right angled")
else:
    print("Triangle is not right angled")
```
